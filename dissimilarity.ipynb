```python
from google.colab import files
import pandas as pd


uploaded = files.upload()


df = pd.read_csv("titanic.csv")

df.head()
```

```python

nominal_df = df[['Sex', 'Survived']].fillna("Unknown")


obj1 = nominal_df.iloc[0]
obj2 = nominal_df.iloc[1]

print("Passenger 1:")
print(obj1)

print("\nPassenger 2:")
print(obj2)

dissimilarity = (obj1 != obj2).sum() / len(obj1)

print("\nNominal Dissimilarity =", dissimilarity)
```

```python
from scipy.spatial.distance import euclidean
import numpy as np


numeric_df = df[['Age', 'Fare']].fillna(df[['Age', 'Fare']].mean())

obj1 = numeric_df.iloc[0]
obj2 = numeric_df.iloc[1]

distance = euclidean(obj1, obj2)

print("Numeric (Euclidean) Dissimilarity =", distance)
```

```python
from scipy.spatial.distance import pdist, squareform

dist_matrix = squareform(pdist(numeric_df, metric='euclidean'))

dist_df = pd.DataFrame(dist_matrix)

print(dist_df.head())
```
