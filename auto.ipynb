```python
from google.colab import files
import pandas as pd


uploaded = files.upload()


df = pd.read_csv("Automobile_data.csv")


print(df.columns.tolist())

df.head()
```

```python

nominal_df = df.select_dtypes(include=['object']).fillna("Unknown")

print("Nominal Columns:")
print(nominal_df.columns.tolist())


obj1 = nominal_df.iloc[0]
obj2 = nominal_df.iloc[1]

dissimilarity = (obj1 != obj2).sum() / len(obj1)

print("\nNominal Dissimilarity =", dissimilarity)
```

```python
from scipy.spatial.distance import euclidean

numeric_df = df.select_dtypes(include=['int64', 'float64'])


numeric_df = numeric_df.fillna(numeric_df.mean())

print("Numeric Columns:")
print(numeric_df.columns.tolist())

obj1 = numeric_df.iloc[0]
obj2 = numeric_df.iloc[1]

distance = euclidean(obj1, obj2)

print("\nNumeric (Euclidean) Dissimilarity =", distance)
```

```python
from scipy.spatial.distance import pdist, squareform

dist_matrix = squareform(pdist(numeric_df, metric='euclidean'))

dist_df = pd.DataFrame(dist_matrix)

print(dist_df.head())
```
