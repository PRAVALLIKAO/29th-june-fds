```python

import pandas as pd
import matplotlib.pyplot as plt


from google.colab import files
uploaded = files.upload()


df = pd.read_csv("Titanic-Dataset (1).csv")


print("First 5 Rows:")
print(df.head())


num_df = df.select_dtypes(include=['number'])

print("\nMeasures of Dispersion:\n")


print("Range:")
print(num_df.max() - num_df.min())


print("\nVariance:")
print(num_df.var())

print("\nStandard Deviation:")
print(num_df.std())

Q1 = num_df.quantile(0.25)
Q3 = num_df.quantile(0.75)
IQR = Q3 - Q1

print("\nInterquartile Range (IQR):")
print(IQR)

num_df.hist(figsize=(12,10), bins=20)
plt.suptitle("Histogram of Numerical Features")
plt.show()


plt.figure(figsize=(12,6))
num_df.boxplot(rot=45)
plt.title("Boxplot of Numerical Features")
plt.show()


num_df.plot(kind='density', subplots=True, layout=(3,3), figsize=(12,10), sharex=False)
plt.suptitle("Density Plots")
plt.show()
```
